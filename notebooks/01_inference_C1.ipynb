```python
from pathlib import Path
import sys, cv2, matplotlib.pyplot as plt
PROJECT = Path.cwd()
if not (PROJECT / "src").is_dir() and (PROJECT.parent / "src").is_dir(): PROJECT = PROJECT.parent
sys.path.insert(0,str(PROJECT))
from src.classification.hair_predictor import HairPredictor
from src.pipeline import process_image
print("Projet :",PROJECT)
```

```python
# Modifier le chemin de l'image : n'importe quelle nouvelle image RGB dermoscopique.
IMAGE_PATH = PROJECT / "test_images" / "example.jpg"
if not IMAGE_PATH.exists():
    print("Ajoutez une image à :", IMAGE_PATH)
else:
    bgr=cv2.imread(str(IMAGE_PATH))
    rgb=cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    detector=HairPredictor(PROJECT, "SVM RBF")
    r=process_image(rgb,detector)
    print("Classes :",r["classes"],"
Détails :",r["details"],"
Traitements :",r["applied"])
    fig,axes=plt.subplots(1,2,figsize=(10,4))
    axes[0].imshow(rgb); axes[0].set_title("Originale")
    axes[1].imshow(r["cleaned_image"]); axes[1].set_title("Nettoyée si C1 positif")
    for ax in axes: ax.axis("off")
    plt.show()
```
